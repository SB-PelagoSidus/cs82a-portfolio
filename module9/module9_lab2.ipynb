# Module 9 Lab 2: Problem 6.3 Documenting a Migration
## Google Colab → campus JupyterHub

## 1. What Transfers Unchanged
For migrating my coursework from Google Colab to campus JupyterHub, I would first download my existing Jupyter notebook in .ipynb format. Since both platforms support Jupyter notebooks, I can transfer the Python code, Markdown documentation, notebook structure, and saved outputs without major changes. However, the computing environment and associated resources may differ between platforms. Therefore, after migration, I would need to execute the notebook and verify that the results can be reproduced correctly.

## 2. What Needs to Be Redone
Although the notebook itself can be transferred, I would need to prepare the new computing environment on campus JupyterHub. This would include obtaining access through my student account, transferring the datasets to the new file storage, and checking the installed Python libraries. Some packages might need to be installed again, and file paths could require modifications because the platforms use different storage locations. After completing these steps, I would run the notebook to verify that everything works correctly.

## 3. One Risk of Migration
One important risk of migrating my coursework from Google Colab to campus JupyterHub is that the notebook might not produce the same results in the new computing environment. Although the Python code remains unchanged, differences in installed libraries, package versions, or file locations could cause errors. To minimize this risk, I would keep a backup of the original notebook, check the software dependencies, and run the notebook again to verify the results.

## 4. My Personal Experience:
My experience moving from local JupyterLab to Google Colab demonstrated that migration involves more than transferring Python code.  During Module 8 labs, I learned that notebooks saved in Google Drive needed to be downloaded separately as local .ipynb files, unlike my previous workflow in local JupyterLab, where I saved notebooks directly on my computer. During Module 9 Lab 1, I encountered a saving conflict in Colab and had to review the changes before saving the correct version. These experiences emphasized the importance of data integrity, maintaining backups, and verifying that the saved notebook contains the correct code and results when working across computing platforms.
