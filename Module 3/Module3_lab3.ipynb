I first analyzed the table provided in Problem 2.2. There are 35 bridges, and length data are missing for 14 of them, which is about 40%. Therefore, I see a problem with the amount of missing data.
```python
flagged = df[(df["length"] <= 0) | (df["length"] > 1600)]
flagged
```

## Review of My Initial Assumption
My initial assumption was based on the textbook table, but the complete CSV showed a broader distribution. Therefore, I need to review my initial assumption after investigating the flagged values.

```python

print("Number flagged:", len(flagged))
print("Percent of available lengths flagged:", round(len(flagged) / df["length"].count() * 100, 1), "%")
print("Flagged length range:", flagged["length"].min(), "to", flagged["length"].max())
```

## Final Conclusion and Handling

My initial assumption of 1600 feet was too restrictive because I did not have all the data. Now I can see that it flagged 30 of the 81 available length values, or about 37%. The summary statistics also show that the median length is 1300 feet, the 75th percentile is 2000 feet, and the maximum is 4558 feet. This means that values greater than 1600 are not unusual enough, and we cannot consider them outliers only because they exceed my initial limit.

After reviewing the flagged rows, I do not think these values should be removed or changed. The largest value, 4558, is unusually high, but it may be reasonable because it is a railroad bridge made of steel and is classified as Long, which supports keeping the value. I would keep the flagged values and document them rather than correct or remove them without additional information.
