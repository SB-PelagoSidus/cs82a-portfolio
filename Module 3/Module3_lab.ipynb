# Initial Data Problems

1. Some ZIP codes have only 4 digits, so a leading zero may be missing.
2. The dates are recorded in different formats.
3. Some prices show NaN, which means the price value is missing.

```python
df.isna().sum()
```

```python
df["price"].median()
```

```python
fill_value = df["price"].median()
print(fill_value)
```

```python
df["price"] = df["price"].fillna(fill_value)
```

```python
df.isna().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.shape
```

```python
df["zip"] = df["zip"].astype(str).str.zfill(5)
```

```python
df["zip"].head(10)
```

```python
df["date"] = pd.to_datetime(df["date"], format="mixed")
```

# Negative Quantity Decision

I saw 3 rows with negative quantities. I desided  tho leave these rows because there is not enough information to conclude that they are data-entry errors. Probably, these negative quantities may represent product returns, and removing or changing them could incorrectly increase the reported net sales quantity.
