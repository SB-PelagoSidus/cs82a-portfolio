# Module 2 Lab 3: Textbook Problem 4.4

Cold: below 15°C. Warm: from 15°C through 25°C, including 15°C and 25°C. Hot: above 25°C.

```python
temperatures = [10, 16, 23, 28, 14, 22, 30]

for temp in temperatures:
    if temp < 15:
        label = "Cold"
    elif temp <= 25:
        label = "Warm"
    else:
        label = "Hot"

    print(temp, "°C:", label)
```

```python
boundary_temperatures = [15, 25]

for temp in boundary_temperatures:
    if temp < 15:
        label = "Cold"
    elif temp <= 25:
        label = "Warm"
    else:
        label = "Hot"

    print(temp, "°C:", label)
    
```
