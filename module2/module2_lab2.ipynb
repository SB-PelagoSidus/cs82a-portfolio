```python
def multiply_for(a, b):
    result = 0
    for i in range(b):
        result = result + a
    return result

print(multiply_for(7, 6))
```

```python
def multiply_while(a, b):
    result = 0
    count = 0
    while count < b:
        result = result + a
        count = count + 1
    return result 

print(multiply_while(7, 6))

```

```python
print(multiply_for(3, 4))
print(multiply_while(3, 4))
```

The cycle continues while the count value is lower than b and stops when the count reaches b.
