```python
import pandas as pd

df = pd.read_csv("sales_clean.csv")
df.shape
```

```python
print(df["price"].mean())
print(df["price"].median())
print(df["price"].std())

print(df["qty"].mean())
print(df["qty"].median())
print(df["qty"].std())
```

The median price is 
37.53 , while the mean price is 55.52. The median is more appropriate because the mean is sensitive to high price values and can be pulled upward by them.

```python
ax = df["price"].plot(kind="hist", bins=20)
ax.set_title("Distribution of Price")
ax.set_xlabel("Price ($)")
ax.set_ylabel("Frequency")
```

The price distribution is right-skewed, with most prices at the lower end and a long tail toward higher prices.
This agrees with my choice of the median as the more appropriate summary for price.

```python
df["revenue"] = df["price"] * df["qty"]
df.head()
```

```python
df.corr(numeric_only=True)
```

```python
ax = df.plot.scatter(x="price", y="revenue")
ax.set_title("Price vs. Revenue")
ax.set_xlabel("Price ($)")
ax.set_ylabel("Revenue")
```

```python
ax = df.plot.scatter(x="qty", y="revenue")
ax.set_title("Quantity vs. Revenue")
ax.set_xlabel("Quantity")
ax.set_ylabel("Revenue")
```

### Analyst's statement

The strongest positive  off-diagonal correlation is between price and revenue, with r = 0.8177. 
The first scatter plot also shows a strong positive linear relationship between them: higher prices are associated with higher revenue. However, several points are far from the main group, for example, prices above 350 dollars with revenue around 1,500–1,700. These points should be checked to determine whether they are outliers.
The second scatter plot shows many individual points located at the same quantity, but with different revenue values. The correlation between quantity and revenue is weaker, with r = 0.3342, which shows a moderate positive relationship. The plot also shows some negative quantity values, which should be investigated to determine whether they are valid observations or possible data errors.
