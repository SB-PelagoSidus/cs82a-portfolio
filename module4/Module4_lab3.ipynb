### Pearson correlation 
The Pearson correlation between list price and best (sale) price is r = 0.9985 (about 0.999) showing an extremely strong positive relationship.


```python
df.plot(
    kind="scatter",
    x="X",
    y="Y",
    title="Higher List Prices Are Associated with Higher Sale Prices",
    xlabel="List Price ($1,000s)",
    ylabel="Best (Sale) Price ($1,000s)"
)
```

### Prediction

I chose a list price of $20,000 (X = 20) because it is within the data range and has nearby data points on both sides.

I used rows 11 and 21 from the table:
- Row 11: X = 20.3 → Y = 17.7
- Row 21: X = 19.7 → Y = 17.2

Since X = 20 is halfway between 19.7 and 20.3, I estimated the best price as halfway between 17.2 and 17.7:

Predicted Y = (17.2 + 17.7) / 2 = 17.45

The predicted best price for a truck with a list price of $20,000 is approximately 17,45 dollars


### Confidence

I have high confidence in this prediction because r is about 0.999, showing an extremely strong positive linear relationship, and 20,000 dollars is within the observed data range of 12,400–22,400 dollars. However, a prediction for a list price above 25,000 dollars would be less reliable because it is outside the observed data range.
