## Housing Type Bar Chart

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

custdata = pd.read_csv("custdata.tsv", sep="\t")
custdata.head()
```

```python
custdata["housing.type"].value_counts(dropna=False)
```

### Clean Missing Housing Data

```python
excluded_rows = custdata["housing.type"].isna().sum()

custdata_clean = custdata.dropna(subset=["housing.type"])

print("Rows excluded:", excluded_rows)
print("Rows remaining:", len(custdata_clean))
```

###  Count and Sort Housing Types


```python
housing_counts = custdata_clean["housing.type"].value_counts().sort_values()
housing_counts
```

###  Create Housing Type Bar Chart

```python
plt.figure(figsize=(10, 5))

sns.barplot(x=housing_counts.index, y=housing_counts.values)

plt.title("Most Customers Have a Mortgage or Rent Their Homes")
plt.xlabel("Housing Type")
plt.ylabel("Number of Customers")
plt.xticks(rotation=25, ha="right")
plt.ylim(bottom=0)
plt.tight_layout()
plt.show()
```

###  Summary
I excluded 56 rows with missing housing-type values (NA), leaving 944 customers distributed across four housing categories for the bar chart.
