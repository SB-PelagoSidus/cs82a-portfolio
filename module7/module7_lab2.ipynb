```python
import pandas as pd
from sklearn.linear_model import LinearRegression
```

## Step 1: Load and inspect the data


```python
df = pd.read_csv("ratings.tsv", sep="\t")

print(df.head())
print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())
```

## Step 2: Fit the regression model

```python
X = df[["ambience", "food", "service"]]
y = df["rating"]

model = LinearRegression()
model.fit(X, y)

print("Intercept:", round(model.intercept_, 4))

for feature, coefficient in zip(X.columns, model.coef_):
    print(feature, ":", round(coefficient, 4))
```

## Step 3: Rank the coefficients

```python
coefficients = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": model.coef_
})

coefficients = coefficients.sort_values(
    by="Coefficient", ascending=False
)

print(coefficients.to_string(index=False))
```

## Step 4: Business recommendation
1. Predicted rating = -15.8713 + 0.1338 × ambience + 0.7338 × food + 0.3240 × service
2. Food has the largest regression coefficient (0.7338), followed by service (0.3240) and ambience (0.1338). A struggling restaurant should prioritize food quality, but these results show associations rather than proving that improvements will cause higher overall ratings.
